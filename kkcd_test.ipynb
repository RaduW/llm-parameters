```python
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm
import os

def load_custom_font(font_filename: str, font_dir: str = os.path.expanduser("~/Library/Fonts")) -> fm.FontProperties:
    """
    Load a specific .ttf font file from a directory and return a FontProperties object.

    Parameters:
        font_filename (str): The TTF filename, e.g., 'ComicNeue-Regular.ttf'
        font_dir (str): Directory containing the font. Default is '~/Library/Fonts'

    Returns:
        FontProperties: The font properties object usable in matplotlib plots.
    """
    font_path = os.path.join(font_dir, font_filename)
    if not os.path.exists(font_path):
        raise FileNotFoundError(f"Font not found at: {font_path}")

    return fm.FontProperties(fname=font_path)

def xkcd_plot_with_font(x_data, y_data, title="", xlabel="", ylabel="", font_prop=None):
    """
    Plot a line chart in XKCD style with optional custom font.

    Parameters:
        x_data (list): X-axis data points.
        y_data (list): Y-axis data points.
        title (str): Plot title.
        xlabel (str): X-axis label.
        ylabel (str): Y-axis label.
        font_prop (FontProperties): Font to apply to labels.
    """
    with plt.xkcd():
        fig, ax = plt.subplots()
        ax.plot(x_data, y_data, marker="o")
        ax.set_title(title, fontproperties=font_prop)
        ax.set_xlabel(xlabel, fontproperties=font_prop)
        ax.set_ylabel(ylabel, fontproperties=font_prop)
        plt.tight_layout()
        plt.show()
```

```python


# Load Comic Neue manually
font = load_custom_font("ComicNeue-Regular.ttf")

# Create a fun XKCD-style plot
xkcd_plot_with_font(
    x_data=[0, 1, 2, 3],
    y_data=[1, 2, 1, 3],
    title="Manual Font XKCD Plot",
    xlabel="Time",
    ylabel="Value",
    font_prop=font
)
```
